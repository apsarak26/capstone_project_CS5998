```python

import pandas as pd
import zipfile

zip_path = "/content/drive/MyDrive/aeolus_v1.zip"

# Column names from the AEOLUS README
schemas = {
    "standard_case_drug.tsv": [
        "primaryid", "isr", "drug_seq",
        "role_cod", "standard_concept_id"
    ],
    "standard_case_outcome.tsv": [
        "primaryid", "isr", "pt",
        "outcome_concept_id", "snomed_outcome_concept_id"
    ],
    "concept.tsv": [
        "concept_id", "concept_name", "domain_id",
        "vocabulary_id", "concept_class_id",
        "standard_concept", "concept_code",
        "valid_start_date", "valid_end_date",
        "invalid_reason"
    ]
}

# Read the first 10 rows of each file
with zipfile.ZipFile(zip_path, "r") as archive:
    for filename, columns in schemas.items():
        with archive.open(filename) as file:
            df = pd.read_csv(
                file,
                sep="\t",
                header=None,
                names=columns,
                nrows=10,
                dtype=str
            )

        print(f"\nFILE: {filename}")
        display(df)
```

```python

with zipfile.ZipFile(zip_path, "r") as archive:
    for filename, columns in schemas.items():
        with archive.open(filename) as file:
            df = pd.read_csv(
                file,
                sep="\t",
                header=None,
                names=columns,
                nrows=10000,
                dtype=str
            )

        print(f"\nFILE: {filename}")
        print("Shape:", df.shape)
        print("\nMissing values:")
        print(df.isnull().sum())
        print("\nDuplicate rows:", df.duplicated().sum())
```
