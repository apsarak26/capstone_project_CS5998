```python

import pandas as pd
import zipfile

zip_path = "/content/drive/MyDrive/aeolus_v1.zip"

# Column names from the AEOLUS README
schemas = {
    "standard_case_drug.tsv": [
        "primaryid", "isr", "drug_seq",
        "role_cod", "standard_concept_id"
    ],
    "standard_case_outcome.tsv": [
        "primaryid", "isr", "pt",
        "outcome_concept_id", "snomed_outcome_concept_id"
    ],
    "concept.tsv": [
        "concept_id", "concept_name", "domain_id",
        "vocabulary_id", "concept_class_id",
        "standard_concept", "concept_code",
        "valid_start_date", "valid_end_date",
        "invalid_reason"
    ]
}

# Read the first 10 rows of each file
with zipfile.ZipFile(zip_path, "r") as archive:
    for filename, columns in schemas.items():
        with archive.open(filename) as file:
            df = pd.read_csv(
                file,
                sep="\t",
                header=None,
                names=columns,
                nrows=10,
                dtype=str
            )

        print(f"\nFILE: {filename}")
        display(df)
```

```python

import pandas as pd
import zipfile

with zipfile.ZipFile(zip_path, "r") as archive:
    with archive.open("standard_case_drug.tsv") as file:
        drug_df = pd.read_csv(
            file,
            sep="\t",
            header=None,
            names=schemas["standard_case_drug.tsv"],
            nrows=10000,
            dtype=str
        )

    with archive.open("standard_case_outcome.tsv") as file:
        outcome_df = pd.read_csv(
            file,
            sep="\t",
            header=None,
            names=schemas["standard_case_outcome.tsv"],
            nrows=10000,
            dtype=str
        )
```

```python

for name, df in [
    ("Drug", drug_df),
    ("Outcome", outcome_df)
]:
    print(f"\n{name} dataset")

    print(
        "Both identifiers missing:",
        (df["primaryid"].isna() & df["isr"].isna()).sum()
    )

    print(
        "Both identifiers present:",
        (df["primaryid"].notna() & df["isr"].notna()).sum()
    )

    print(
        "Only primaryid:",
        (df["primaryid"].notna() & df["isr"].isna()).sum()
    )

    print(
        "Only isr:",
        (df["primaryid"].isna() & df["isr"].notna()).sum()
    )
```

```python

def create_report_id(df):
    df = df.copy()

    df["report_id"] = None

    primary_mask = df["primaryid"].notna()
    isr_mask = df["isr"].notna()

    df.loc[primary_mask, "report_id"] = (
        "PRIMARY_" + df.loc[primary_mask, "primaryid"]
    )

    df.loc[~primary_mask & isr_mask, "report_id"] = (
        "ISR_" + df.loc[~primary_mask & isr_mask, "isr"]
    )

    return df

drug_df = create_report_id(drug_df)
outcome_df = create_report_id(outcome_df)

print("Unique drug report IDs:", drug_df["report_id"].nunique())
print("Unique outcome report IDs:", outcome_df["report_id"].nunique())

common_ids = set(drug_df["report_id"].dropna()) & set(
    outcome_df["report_id"].dropna()
)

print("Matching report IDs:", len(common_ids))
```

```python
#Missing value handling
import pandas as pd
import zipfile

with zipfile.ZipFile(zip_path, "r") as archive:
    with archive.open("standard_case_drug.tsv") as file:
        drug_df = pd.read_csv(
            file,
            sep="\t",
            header=None,
            names=schemas["standard_case_drug.tsv"],
            nrows=10000,
            dtype=str,
            na_values=["\\N"]
        )

    with archive.open("standard_case_outcome.tsv") as file:
        outcome_df = pd.read_csv(
            file,
            sep="\t",
            header=None,
            names=schemas["standard_case_outcome.tsv"],
            nrows=10000,
            dtype=str,
            na_values=["\\N"]
        )
```

```python

for name, df in [
    ("Drug", drug_df),
    ("Outcome", outcome_df)
]:
    print(f"\n{name} dataset")

    print(
        "Both missing:",
        (df["primaryid"].isna() & df["isr"].isna()).sum()
    )

    print(
        "Both present:",
        (df["primaryid"].notna() & df["isr"].notna()).sum()
    )

    print(
        "Only primaryid:",
        (df["primaryid"].notna() & df["isr"].isna()).sum()
    )

    print(
        "Only isr:",
        (df["primaryid"].isna() & df["isr"].notna()).sum()
    )
```

```python

drug_df = create_report_id(drug_df)
outcome_df = create_report_id(outcome_df)

print("Unique drug report IDs:", drug_df["report_id"].nunique())
print("Unique outcome report IDs:", outcome_df["report_id"].nunique())

common_ids = (
    set(drug_df["report_id"].dropna())
    & set(outcome_df["report_id"].dropna())
)

print("Matching report IDs:", len(common_ids))
```

```python

# Join drug and outcome samples using report_id
joined_df = pd.merge(
    drug_df,
    outcome_df,
    on="report_id",
    how="inner",
    suffixes=("_drug", "_outcome")
)

print("Joined dataset shape:", joined_df.shape)

display(
    joined_df[
        [
            "report_id",
            "standard_concept_id",
            "role_cod",
            "pt",
            "outcome_concept_id"
        ]
    ].head(10)
)
```

```python

# Explore unique adverse event terms
unique_terms = (
    outcome_df["pt"]
    .dropna()
    .drop_duplicates()
    .sort_values()
)

print("Unique adverse event terms:", len(unique_terms))

# Exploratory search only, not final classification
candidate_terms = unique_terms[
    unique_terms.str.contains(
        "allerg|hypersensitiv|anaphyla|urticaria|angioedema",
        case=False,
        regex=True
    )
]

display(candidate_terms.reset_index(drop=True))
```

```python

import pandas as pd
import zipfile

all_terms = set()
total_rows = 0

with zipfile.ZipFile(zip_path, "r") as archive:
    with archive.open("standard_case_outcome.tsv") as file:
        for chunk in pd.read_csv(
            file,
            sep="\t",
            header=None,
            names=schemas["standard_case_outcome.tsv"],
            dtype=str,
            na_values=["\\N"],
            chunksize=100000
        ):
            total_rows += len(chunk)
            all_terms.update(chunk["pt"].dropna().unique())

print("Total outcome records:", total_rows)
print("Unique adverse event terms:", len(all_terms))
```

```python
#Identify candidate allergy-related terms
terms_df = pd.DataFrame(
    sorted(all_terms),
    columns=["pt"]
)

pattern = (
    "allerg|hypersensitiv|anaphyla|"
    "urticaria|angioedema"
)

candidate_terms_df = terms_df[
    terms_df["pt"].str.contains(
        pattern,
        case=False,
        na=False
    )
].copy()

print("Candidate terms:", len(candidate_terms_df))

display(candidate_terms_df)
```

```python

candidate_terms_df.to_csv(
    "/content/allergy_candidate_terms.csv",
    index=False
)

print("Candidate terms exported successfully.")
```
